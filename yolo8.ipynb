```python
import os
import glob
import shutil
import yaml
from ultralytics import YOLO

BASE_DIR = r"C:\Users\ThinkPad\Final_Project"
DATA_DIR = os.path.join(BASE_DIR, "Data", "construction-ppe")
yaml_path = os.path.join(DATA_DIR, "data.yaml")

# 1. Fix the 'path' inside data.yaml so YOLO finds train/val images without failing
if os.path.exists(yaml_path):
    with open(yaml_path, "r", encoding="utf-8") as f:
        cfg = yaml.safe_load(f)
    cfg["path"] = DATA_DIR
    with open(yaml_path, "w", encoding="utf-8") as f:
        yaml.safe_dump(cfg, f)
    print("Updated data.yaml path to:", DATA_DIR)
else:
    raise FileNotFoundError(f"data.yaml not found at {yaml_path}")

# 2. Train YOLOv8n (using 5 epochs & imgsz=416 first so it finishes quickly on CPU)
model = YOLO("yolov8n.pt")
results = model.train(
    data=yaml_path,
    epochs=5,
    imgsz=416,
    batch=8,
    project=os.path.join(BASE_DIR, "runs"),
    name="stage1_yolo",
    exist_ok=True
)

# 3. Copy the newly generated best.pt directly into C:\Users\ThinkPad\Final_Project\best.pt
expected_best = os.path.join(BASE_DIR, "runs", "stage1_yolo", "weights", "best.pt")
final_best = os.path.join(BASE_DIR, "best.pt")

if os.path.exists(expected_best):
    shutil.copy(expected_best, final_best)
    print(f"SUCCESS! best.pt is now ready at: {final_best}")
else:
    # Search anywhere on your user folder in case Ultralytics saved it elsewhere
    found = glob.glob(r"C:\Users\ThinkPad\**\best.pt", recursive=True)
    print("Could not find at expected path. Found best.pt files here:", found)
```
